# UNION Y LIMPIEZA

```python
#junto el df de eventos y por facturacion por `id_booking` para ver de golpe qué peticiones se ganaron y cuánto facturó cada una.
# Uso how='outer' para no perder nada: ni las peticiones perdidas, ni los 23 eventos metidos sin booking file
```

```python
df = eventos.merge(facturacion, on='id_booking', how='outer')
```

```python
print(df.shape)
```

```python
df.head()
```

```python
df['fecha_inicio'] = df['fecha_inicio_x'].fillna(df['fecha_inicio_y'])
df['cliente'] = df['cliente'].fillna(df['nombre_cliente'])
df = df.drop(columns=['fecha_inicio_x', 'fecha_inicio_y', 'nombre_cliente'])
```

```python
df.head()
```

```python
df = df[['id_booking', 'fecha_inicio', 'producto', 'estado', 'cliente',
         'fecha_fin', 'fecha_creacion', 'tipo_evento', 'pax',
         'total_evento', 'sala', 'fb', 'alojamiento']]
```

```python
df.head()
```

```python
df.tail()
```

```python
df[['total_evento','sala','fb','alojamiento']] = df[['total_evento','sala','fb','alojamiento']].fillna(0)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['producto'] = df['producto'].fillna('Salas')
```

```python
df.head()
```

```python

((df['total_evento'] == 0) & (df['estado'] == 'CONFIRMACIÓN')).sum() 
```

```python
df[(df['total_evento'] == 0) & (df['estado'] == 'CONFIRMACIÓN')]
```

```python
df['estado'] = df['estado'].fillna('CONFIRMACIÓN')
```

```python
df.head()
```

```python
df['antelacion_dias'] = (df['fecha_inicio'] - df['fecha_creacion']).dt.days
```

```python
df.head()
```

```python
df.tail()
```

```python
df['estado'].value_counts()
```

```python
df['estado'] = df['estado'].replace('Rechazado', 'Perdido')
df = df.drop(columns='cliente')
```
